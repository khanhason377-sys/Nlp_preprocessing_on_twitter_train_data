```python
def text_preprocessor(text):
    text = text.lower()
    text = re.sub("\s+"," ",text)
    text = re.sub("\d+"," ",text)
    text = "".join([a for a in text if a not in string.punctuation])
    text = " ".join([a for a in text.split() if a not in stopwords.words("english")])
    return text
```
